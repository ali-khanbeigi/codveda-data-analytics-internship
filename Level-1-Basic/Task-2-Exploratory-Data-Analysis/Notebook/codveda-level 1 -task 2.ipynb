```python
summary_stats = pd.DataFrame({
    "Mean": numeric_cols.mean(),
    "Median": numeric_cols.median(),
    "Mode": numeric_cols.mode().iloc[0],
    "Standard Deviation": numeric_cols.std()
})

summary_stats
```

```python
df.describe()
```

```python
### Observation

#The descriptive statistics show noticeable differences in the distributions
# of the Iris features. Petal length has the highest standard deviation,
# indicating greater variability among the observations. Sepal width has
# the lowest standard deviation and therefore shows less variation.
```

```python
### Observation

#Sepal length and sepal width show relatively continuous distributions.
#Petal length and petal width exhibit more distinct groups, suggesting
#differences between the Iris species.
```

```python
sns.histplot(
    data=df,
    x = "petal_length",
    hue = "species",
    kde = True
)
plt.title("Petal Length Distribution by Species")
plt.xlabel("Petal Length")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig(r"D:\projects\codveda project\level 1-task 2\plots\Petal Length Distribution by Species", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize = (10,6))

sns.boxplot(data = numeric_cols )

plt.title("Boxplot of Numerical Features")
plt.xlabel("Features")
plt.ylabel("Value")
plt.tight_layout()
plt.savefig(r"D:\projects\codveda project\level 1-task 2\plots\Boxplot of Numerical Features", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.boxplot(
    data = df,
    x = "species",
    y = "petal_length"
)
plt.title("Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length")
plt.tight_layout()
plt.savefig(r"D:\projects\codveda project\level 1-task 2\plots\Petal Length by Species", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.boxplot(
    data=df,
    x="species",
    y="petal_width"
)

plt.title("Petal Width by Species")
plt.xlabel("Species")
plt.ylabel("Petal Width")
plt.tight_layout()
plt.savefig(r"D:\projects\codveda project\level 1-task 2\plots\Petal Width by Species", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="petal_length",
    y="petal_width",
    hue="species"
)

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.tight_layout()
plt.savefig(r"D:\projects\codveda project\level 1-task 2\plots\Petal Length vs Petal Width", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="sepal_length",
    y="sepal_width",
    hue="species"
)

plt.title("Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.tight_layout()
plt.savefig(r"D:\projects\codveda project\level 1-task 2\plots\Sepal Length vs Sepal Width", dpi=300, bbox_inches="tight")
plt.show()
```

```python
### Observation

#Petal length and petal width show a strong positive relationship.
#The species also form relatively distinct groups, particularly Setosa,
#which is clearly separated from Versicolor and Virginica.
```

```python
df.groupby("species")[["sepal_length", "sepal_width", "petal_length", "petal_width"]].mean()
```

```python
### Observation

#Petal measurements show clear differences among the three Iris species.
#Setosa has the smallest petals, Versicolor has intermediate petal sizes,
#and Virginica has the largest petals. This suggests that petal length and
#petal width may be useful features for distinguishing between the species.
```

```python
correlation = numeric_cols.corr()
correlation
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix of Iris Features")
plt.tight_layout()
plt.savefig(r"D:\projects\codveda project\level 1-task 2\plots\Sepal Length vs Sepal Width", dpi=300, bbox_inches="tight")
plt.show()
```

```python
### Correlation Analysis

#The correlation matrix shows a very strong positive relationship between
#petal length and petal width (0.96). Sepal length also has strong positive
#correlations with petal length (0.87) and petal width (0.82).

#In contrast, sepal width has moderate negative correlations with petal
#length (-0.42) and petal width (-0.36). The correlation between sepal
#length and sepal width is weak (-0.11).

#Overall, petal length and petal width are the most strongly related
#features in the dataset.
```

```python
## Conclusion

#The exploratory analysis of the Iris dataset revealed clear differences between the three flower species.

#Petal length and petal width showed the strongest ability to distinguish
#between species, with Setosa having the smallest petals and Virginica having the largest.

#A strong positive correlation (0.96) was observed between petal length and petal width.
#Sepal measurements showed weaker separation between species compared with petal measurements.

#Overall, petal-related features appear to provide the most useful information for differentiating Iris species.
```
